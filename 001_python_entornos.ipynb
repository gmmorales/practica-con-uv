## Antes de abrir el cuaderno

Estos pasos se realizan **una sola vez, desde la terminal de VS Code**, ubicada en la carpeta de esta clase. Todavía no se ejecutan en una celda porque aún no existe el entorno que ejecutará el cuaderno.

### 1. Instalar `uv`

En Windows PowerShell:

```powershell
powershell -ExecutionPolicy ByPass -c "irm https://astral.sh/uv/install.ps1 | iex"
uv --version
```

Si la terminal no reconoce `uv`, cerrarla y abrirla nuevamente. Para otros sistemas operativos, consultar [la instalación oficial de uv](https://docs.astral.sh/uv/getting-started/installation/).

### 2. Crear el proyecto y su entorno con Python 3.12

```powershell
uv init --bare --python 3.12
uv venv --python 3.12
uv add numpy matplotlib pillow gradio
uv add --dev ipykernel
```

La opción `--python 3.12` es la instrucción directa: le indica a `uv` qué versión debe usar para crear `.venv`. Si esa versión no está disponible, `uv` puede descargarla.

La elección persistente queda escrita en `.python-version`. También puede establecerse explícitamente:

```powershell
uv python pin 3.12
```

Después de fijarla, `uv venv` puede leer esa versión sin repetir `--python 3.12`.

### 3. Registrar este entorno como kernel

```powershell
uv run python -m ipykernel install --user --name pdi-c02-python312 --display-name "PDI C02 - Python 3.12"
```

- `--name` es el identificador interno y no lleva espacios.
- `--display-name` es el nombre que veremos en VS Code.

Abrir este archivo, pulsar **Seleccionar kernel** y elegir **PDI C02 - Python 3.12**.
